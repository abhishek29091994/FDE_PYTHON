## Data Type

#### int , float, complex, bool,str,list,tuple,range,dic,set

```python
age=25
count=10
temp=-10
print(type(age).__name__)
```

```python
number=25.0
print(type(number).__name__)
```

```python
#complec number
complex_number=2+3j
print(type(complex_number).__name__)
```

```python
name= "sagar"
print(type(name).__name__)
print(len(name))
```

```python
is_student= True
print(type(is_student).__name__)
```

```python
x="10"
y=int(x)
print(type(y).__name__)
print(y+2)
```

```python
# duplicate not allowed

x={1,2,3,4,5}
x.add(6)
print(x)
print(type(x).__name__)       
```

```python
x={
    "apple": 1,
    "banana": 2,   
    "cherry": 3,
    "date": 4
}

x["elderberry"] = 5
print(x)

print(type(x).__name__)
```

```python
#list
list=[1,2,3,4,5]
for i in range(0,4):
    print(list[i])
print(type(list).__name__)
```

```python
#tuple
my_tuple = (1, 2, 3, 4, 5)

for i in range(0, 4):
    print(my_tuple[i])

print(type(my_tuple).__name__)
del list
# Convert tuple into list
my_list = list(my_tuple)

print(my_list)
print(type(my_list).__name__)
```

```python
#range
my_range = range(1, 6)
for i in my_range:
    print(i)
print(type(my_range).__name__)
```
